### Which convention does each calculation use?

**Day-count basis** measures interest; **Street yield** discounts the resulting cashflows. `schedule rule` determines date-generation direction, and payment conventions roll payment dates. These are separate choices.

| Accrual basis | Year fraction for an irregular coupon | Example |
|---|---|---|
| `ACT/ACT (ICMA)` | Sum `(1/f) × overlap days / reference-period days` | HMLGLO first coupon: `4 × (58/181 + 1) = 5.281768` per 100 |
| `30/360 (BOND BASIS)` | Convention-specific 30/360 day count divided by 360 | META first coupon: `100 × 0.0525 × 191/360 = 2.785417` |
| `ISMA-30/360` | Same BondBasis mapping in this library | KSA final coupon: `100 × 0.0375 × 168/360 = 1.75` |
| `ACT/360` | Actual calendar days divided by 360 | A 45-day stub at 6% on face 100 pays `100 × 0.06 × 45/360 = 0.75` |
| `ACT/365 FIXED` | Actual calendar days divided by 365 | The same stub pays `100 × 0.06 × 45/365 = 0.739726` |

Read `first period accrual basis` for the first coupon, `last period accrual basis` for the last, and `accrual basis` for interior periods. These fields can differ. The introductory quarterly example uses **ACT/ACT (ICMA) + Street discounting**.

### Example: schedule direction and payment-date adjustment

A bond starts accruing on **January 15, 2026**, matures on
**November 15, 2026**, and pays coupons every **6 months**.
Assume no first or last regular coupon dates are explicitly specified.

| Schedule rule | Accrual dates, in chronological order | Irregular period |
|---|---|---|
| `FORWARD` | January 15 → July 15 → November 15 | Short final period: 4 months |
| `BACKWARD` | January 15 → May 15 → November 15 | Short first period: 4 months |

`FORWARD` builds the schedule from the accrual start; `BACKWARD` builds it
from maturity. Choose the direction that matches the contractual schedule.

In **both cases**, maturity falls on **Sunday, November 15, 2026**.
With the `Following` payment convention, assuming Monday is a business day,
the final payment moves to **Monday, November 16**.

The final accrual period still ends on **November 15**.
Only the payment date moves; the extra day does not increase coupon accrual.

## 2. Price the same bond

Follow the same array-based steps as the lecture example:
1. Collect the remaining coupon amounts. Express each remaining interval in coupon-period units, then use np.cumsum to obtain cumulative times from settlement.
2. Compute Street discount factors: `(1 + y/f)**(-n)`, or `1/(1 + y*n/f)` for one remaining payment in this model.
3. Compute `coupon_pv = np.dot(coupon_amounts, discount)` and add the discounted redemption.
4. Subtract accrued interest to obtain clean price.

The supplied Bloomberg yields require the Street factors above. Accrued interest is `coupon_amount * elapsed_accrual / full_period_accrual`.

The six implementation tasks have these responsibilities:
- `_generate_bond_schedule`: generate accrual/payment dates; label first, regular and last periods, and assign each period's accrual basis and regular/stub flag.
- `_first_period_accrual`: use the regular method for a normal first period; otherwise build reference periods backward and calculate the first coupon using its accrual basis. Split ICMA stubs where needed; save reference dates and the full-period year fraction.
- `_last_period_accrual`: use the regular method for a normal final period; otherwise build reference periods forward and calculate the final coupon using its accrual basis. Split ICMA stubs where needed; save reference dates and the full-period year fraction. Exclude principal.
- `_regular_period_accrual`: save reference dates and the full-period year fraction; return face × annual coupon rate / frequency.
- `_accrued_interest`: find the current period; return its coupon × elapsed year fraction / full-period year fraction.
- `yield_to_price`: discount remaining coupons and redemption, then subtract accrued interest; also return the first and second price derivatives with respect to yield.

## 4. Additional validation

The Bloomberg assertions above pin five real bonds. They leave three things unchecked,
each of which a wrong implementation can still pass:

1. **Schedule direction and payment rolling.** Every graded case uses `BACKWARD`, so
   nothing tests `FORWARD`, and none has a maturity falling on a weekend. Section A
   reproduces the worked example in the markdown above: the same bond generated both
   ways, with maturity on Sunday 2026-11-15 rolling to Monday for *payment* while the
   final *accrual* date stays put.
2. **Fixed-denominator stubs.** The graded stubs are ICMA and 30/360 only. Section B
   prices a 45-day final stub under `ACT/360` and `ACT/365 FIXED` against the two
   closed-form values in the convention table.
3. **The yield derivatives.** No assertion touches `dpricedyield` or `d2pricedyield2`
   directly — the price tolerance merely scales with the first one. Section D compares
   both against central differences of the dirty price.

Sections C, E and F add three internal consistency checks: that the provided
`price_to_yield` solver inverts `yield_to_price`, that accrued interest vanishes at both
ends of a period, and that the coupons sum to `face x rate x` total accrual.

On step size: the second derivative is differenced at `h = 1e-4` rather than `1e-6`.
A second central difference divides by `h**2`, so at `1e-6` the `~1e-13` relative
rounding error in a price near 100 is amplified to roughly `1e-4` relative error in the
result -- large enough to look like a bug when nothing is wrong. `1e-4` balances that
against the `O(h**2)` truncation error and lands near `1e-7`.

```python
from fixedincomelib import Date

BASE = {'issue date': '2026-01-15', 'first accrual date': '2026-01-15',
        'first coupon date': None, 'maturity date': '2026-11-15',
        'coupon rate': 0.06, 'face value': 100.0, 'redemption': 1.0,
        'coupon accrual period': '6M', 'accrual basis': 'ACT/ACT (ICMA)',
        'first period accrual basis': 'ACT/ACT (ICMA)',
        'last period accrual basis': 'ACT/ACT (ICMA)',
        'schedule rule': 'FORWARD', 'last regular coupon date': None, 'end of month': False,
        'settlement business day convention': 'F', 'settlement holiday convention': 'USGS',
        'settlement offset': '1D', 'payment business day convention': 'F',
        'payment holiday convention': 'USGS', 'payment offset': '0D'}


def mk(**kw):
    """A bond_convention built off BASE; `basis` sets all three accrual-basis fields."""
    d = dict(BASE)
    if 'basis' in kw:
        b = kw.pop('basis')
        d.update({'accrual basis': b, 'first period accrual basis': b,
                  'last period accrual basis': b})
    d.update(kw)
    return d


CASES = [
    ('CT10', {'issue date': '2026-08-17', 'first accrual date': '2026-08-15',
              'first coupon date': '2027-02-15', 'maturity date': '2036-08-15',
              'coupon rate': 0.04625, 'basis': 'ACT/ACT (ICMA)'}, '2026-09-21', 0.04997112),
    ('META', {'issue date': '2026-05-04', 'first accrual date': '2026-05-04',
              'first coupon date': '2026-11-15', 'maturity date': '2036-05-15',
              'coupon rate': 0.0525, 'basis': '30/360 (BOND BASIS)'}, '2026-05-06', 0.05921383),
    ('HMLGLO', {'issue date': '2025-06-18', 'first accrual date': '2025-06-18',
                'first coupon date': '2026-02-15', 'maturity date': '2030-02-15',
                'coupon rate': 0.08, 'basis': 'ACT/ACT (ICMA)'}, '2025-09-15', 0.079858),
]

print('=== A. schedule direction and payment rolling ===')
for rule, dates_e, reg_e in [('FORWARD', ['2026-01-15', '2026-07-15', '2026-11-15'], [True, False]),
                             ('BACKWARD', ['2026-01-15', '2026-05-15', '2026-11-15'], [False, True])]:
    fcd = '2026-11-15' if rule == 'FORWARD' else None
    cf = BondCalculator(bond_convention=mk(**{'schedule rule': rule,
                                              'first coupon date': fcd})).schedule_dataframe()
    dates = list(cf['StartDate']) + [cf.iloc[-1]['EndDate']]
    print(f'  {rule:9s} accrual dates {dates}')
    print(f'            irregular flags {[not r for r in cf["IsRegular"]]}, '
          f'final accrual ends {cf.iloc[-1]["EndDate"]}, pays {cf.iloc[-1]["PaymentDate"]}')
    assert dates == dates_e, f'{rule} generated the wrong accrual dates'
    assert list(cf['IsRegular']) == reg_e, f'{rule} mislabelled the stub'
    assert cf.iloc[-1]['PaymentDate'] == '2026-11-16', 'Sunday maturity did not roll for payment'
    assert cf.iloc[-1]['EndDate'] == '2026-11-15', 'payment rolling moved an accrual date'

print('\n=== B. ACT/360 and ACT/365 FIXED stubs (45-day final period, 6% on face 100) ===')
for basis, want in [('ACT/360', 100 * .06 * 45 / 360), ('ACT/365 FIXED', 100 * .06 * 45 / 365)]:
    cf = BondCalculator(bond_convention=mk(
        basis=basis, **{'maturity date': '2026-06-29', 'schedule rule': 'BACKWARD',
                        'last regular coupon date': '2026-05-15',
                        'first coupon date': None})).schedule_dataframe()
    got = cf.iloc[-1]['Coupon']
    print(f'  {basis:14s} stub coupon {got:.9f}, expected {want:.9f}')
    assert abs(got - want) < 1e-12, f'{basis} stub coupon is wrong'

print('\n=== C. price_to_yield inverts yield_to_price ===')
for name, kw, settle, y in CASES:
    c = BondCalculator(bond_convention=mk(**kw))
    px = c.yield_to_price(settlement_date=settle, yield_rate=y)
    back = c.price_to_yield(settlement_date=settle, price=px['clean_price'], clean=True)
    print(f'  {name:7s} y={y:.8f} -> clean={px["clean_price"]:.8f} -> y={back["yield"]:.12f}'
          f'   round-trip error {abs(back["yield"] - y):.2e}')
    assert abs(back['yield'] - y) < 1e-10, f'{name} failed to round-trip'

print('\n=== D. analytic yield derivatives vs central differences ===')
h1, h2 = 1e-6, 1e-4
for name, kw, settle, y in CASES:
    c = BondCalculator(bond_convention=mk(**kw))
    base = c.yield_to_price(settlement_date=settle, yield_rate=y)

    def dirty(shift):
        return c.yield_to_price(settlement_date=settle, yield_rate=y + shift)['dirty_price']

    fd1 = (dirty(h1) - dirty(-h1)) / (2 * h1)
    fd2 = (dirty(h2) - 2 * base['dirty_price'] + dirty(-h2)) / h2 ** 2
    e1 = abs(base['dpricedyield'] - fd1) / abs(fd1)
    e2 = abs(base['d2pricedyield2'] - fd2) / abs(fd2)
    print(f'  {name:7s} dP/dy   {base["dpricedyield"]:14.6f} vs {fd1:14.6f}   rel {e1:.2e}')
    print(f'          d2P/dy2 {base["d2pricedyield2"]:14.6f} vs {fd2:14.6f}   rel {e2:.2e}')
    assert e1 < 1e-7, f'{name} first derivative disagrees with the finite difference'
    assert e2 < 1e-6, f'{name} second derivative disagrees with the finite difference'

print('\n=== E. accrued interest at both ends of a period ===')
c = BondCalculator(bond_convention=mk(**CASES[0][1]))
row = c.schedule_dataframe().iloc[3]
ai_start = c._accrued_interest(Date(row['StartDate']))
ai_end = c._accrued_interest(Date(row['EndDate']))
print(f'  period {row["StartDate"]} -> {row["EndDate"]}, coupon {row["Coupon"]:.6f}')
print(f'  AI on the start date {ai_start:.12f}; AI on the end date {ai_end:.12f}')
print('  (both zero: settlement on an accrual end date rolls into the next period, '
      'where nothing has accrued)')
assert ai_start == 0.0 and ai_end == 0.0, 'accrued interest is nonzero at a period boundary'

print('\n=== F. coupons reconcile with the saved accrual factors ===')
for name, kw, settle, y in CASES:
    c = BondCalculator(bond_convention=mk(**kw))
    cf = c.schedule_dataframe()
    got, want = cf['Coupon'].sum(), c.face_value * c.coupon_rate * cf['Accrued'].sum()
    print(f'  {name:7s} sum(Coupon) {got:.10f} vs face x rate x sum(Accrued) {want:.10f}')
    assert abs(got - want) < 1e-9, f'{name} coupons do not reconcile with the accrual factors'

print('\nAll additional checks passed.')
```
